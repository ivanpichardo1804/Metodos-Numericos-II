#Codigo de integración por Simpson compuesto

Importamos la librería Sympy

```python
import sympy as sp #importamos la libreria
```

Se registran los valores principales para la iteración

```python
b = input("Ingrese el valor de b:")
a = input("Ingrese el valor de a:")
n = input("Ingrese el valor de n:")
b = int(b)
a = int(a)
n = int(n)

h = (b-a)/n
```

Aqui hacemos que el programa esté preparado para recibir cualquier función en forma de texto con ayuda de sympy

```python
x = sp.Symbol('x') #asignamos la variable aleatoria X

entrada = input("Ingresa tu función en términos de x (ej. x**2 + exp(x)): ")   #La función de ingresa en forma escrita

expresion = sp.sympify(entrada) #traducimos el texto para que programa pueda comprender que función se va a utilizar
fx = sp.lambdify(x, expresion, 'math') #se define la función matemática
```

Empezamos a desarrollar el método de integración de Simpson compuesto

```python
XI0 = fx(a) + fx(b) #dado que f(b) y f(a) van por defecto, las ingresamos como valores pre-establecidos
```

Vamos calculando el valor de cada f(xi) en los intervalos

```python
lista_X = {
    "XI0": XI0,
} #este diccionario contendrá todos los valores f(xi), y se usa diccionario para poder llamarlos de forma más literal o explícita a como se haría a papel
i=1
while i<=n-1:
  X = a + i*h #esto es para dividir los intervalos del rango de a-b
  lista_X[f"XI{i}"] = fx(X)
  i+=1
```

Se hace la sumatoria completa para estimar el valor de la función según el método

```python
i = 1
XI = XI0
while i<=n-1:
  if i%2 == 0: #pares
    XI += 2*lista_X[f"XI{i}"]
  else:   #impares
    XI += 4*lista_X[f"XI{i}"]
  i+=1

XI_final = (XI*h)/3

print(XI_final)
```

# Definicion de la funcion para ser llamada y utilizada más fácilmente

```python
def simpson_compuesto(a, b, n, funcion):
  b = int(b)
  a = int(a)
  n = int(n)

  h = (b-a)/n

  x = sp.Symbol('x')

  entrada = str(funcion)

  expresion = sp.sympify(entrada)
  fx = sp.lambdify(x, expresion, 'math')

  XI0 = fx(a) + fx(b)

  lista_X = {
    "XI0": XI0,
  }
  i=1
  while i<=n-1:
    X = a + i*h
    lista_X[f"XI{i}"] = fx(X)
    i+=1

  i = 1
  XI = XI0
  while i<=n-1:
    if i%2 == 0: #pares
      XI += 2*lista_X[f"XI{i}"]
    else:   #impares
      XI += 4*lista_X[f"XI{i}"]
    i+=1

  XI_final = (XI*h)/3

  return XI_final
```

```python
simpson_compuesto(1,4,6, "3*x**2 + 5*x + sin(x)")
```
