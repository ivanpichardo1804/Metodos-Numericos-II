```python
import sympy as sp
```

```python
x = sp.Symbol('x') #asignamos la variable aleatoria X

entrada = input("Ingresa tu función en términos de x (ej. x**2 + exp(x)): ")   #La función se ingresa en forma escrita

expresion = sp.sympify(entrada) #traducimos el texto para que programa pueda comprender que función se va a utilizar
fx = sp.lambdify(x, expresion, 'math') #se define la función matemática
```

```python
a = int(input("Ingresa el límite inferior: "))
b = int(input("Ingresa el límite superior: "))
epsilon = float(input("Ingresa el límite superior: "))
```

Usando la formula de Simpson que se define como
$$
\int_a^b f(x) = S(a,b) = [ f(a) + 4 \cdot f(a+h) + f(b)]
$$

```python
h1 = (b - a) / 2

iteracion_1 = (fx(a) + 4*fx(a+h1) + fx(b)) * (h1 / 3)
error = 1 # asignamnos un valor aleatorio al error solo para poder empezar el ciclo while de adelante
salto = 2*h1 #empezamos el salto desde aqui para evitar confusiones
resultados = [iteracion_1] #listas donde se almacenan todos los resultados de las iteraciones para comparar despues
n = 0 #este contador servira para contar el numero de iteraciones y como punto de referencia para la comparacion de resultados en el error

while error >= epsilon: #esto solo es la condicion de precision
    salto = salto / 2  # reducimos el salto una sola vez por iteración
    lista = [] # aqui almacenaremos todos los valores por cada iteracion
    i = a # empezamos siempre en 1, cada vez que la iteracion anterior no haya sido suficiente

    while (i + salto) <= b: # antes de que i alcance a b
        lista.extend([fx(i), 4*fx(i + (salto / 2)), fx(i + salto)]) # añadimos cada elemento a la lista
        i += salto

    iteracion = (sum(lista)) * ((salto/2) / 3) # se saca el resultado de la iteracion
    resultados.append(iteracion)

    error = abs((resultados[n+1] - resultados[n])) / 15
    print(f"Error en iteración {n+1}: {error}")

    n += 1 #para pasar a la siguiente iteracion

print(f"El resultado final de la iteración sería {resultados[n]} con {n} iteraciones.")
```

El libro indica que el resultado deberia ser aproximado de $-1.426014$ por lo que la estimacion es bastante buena.
