# Codigo de integracion por Romberg

Importamos la librería Sympy

```python
import sympy as sp
```

Se registran los valores principales para la iteración

```python
a = input("Ingrese valor de a: ")
b = input("Ingresé valor de b: ")
n = input("Ingresé el valor de hasta que Rn,n busca llegar: ")

a = float(a)
b = float(b)
n = int(n)
```

Aqui hacemos que el programa esté preparado para recibir cualquier función en forma de texto con ayuda de sympy

```python
x = sp.Symbol('x') #asignamos la variable aleatoria X

entrada = input("Ingresa tu función en términos de x (ej. x**2 + exp(x)): ")   #La función se ingresa en forma escrita

expresion = sp.sympify(entrada) #traducimos el texto para que programa pueda comprender que función se va a utilizar
fx = sp.lambdify(x, expresion, 'math') #se define la función matemática
```

Para calcular R11

```python
#Este diccionario contendrá todas las columnas Rj,k, y se usa diccionario para poder llamarlos de forma más literal o explícita a como se haría a papel
lista_R = {
    "R11" : ((b-a)/2) * (fx(a) + fx(b))
}
```

```python
lista_R
```

Definimos una funcion para hj ya que puede llegar a confundirse facilmente entre tantos terminos y variando en el valor de j

```python
def h(j):
  return (b-a) / 2**(float(j)-1)
```

Para R21, R31, R41, .... usamos la regla compuesta trapezoidal

```python
i=2

while i<=n:
  lista_R[f"R{i}1"] = lista_R[f"R{i-1}1"] / 2   #este termino seria: 1/2 Rj-1,1
  sumatoria=0
  k=1
  while k <= 2**(i-2):    # este ciclo es para calcular la serie/sumatoria que empieza en k=1 hasta 2^j-2
    sumatoria += fx(a + (2*k-1)* h(i))    #este termino seria f(a + (2i-1)hj )
    k+=1
  lista_R[f"R{i}1"] += 0.5 * h(i-1) * sumatoria   #completamos Rj,1 con la sumatoria y los valores que conseguimos antes
  i+=1
```

```python
lista_R #Esto solo sirve para corroborar que los valores vayan por buen camino o para ir detectando donde podría estar el error
```

Para R22, R32, R42,
R33, R43,
R44, ...

```python
i=2
j=2 #Reiniciamos los contadores, ya que terminaron con el ultimo valor utlizado del cuadro de arriba

while j<=n:
  i=j   #esta igualdad es muy importante para que al terminar una columna, pueda empezar con la siguiente (ya que se resetean los indices al empezar cada columna)
  print(i,j)
  while i<=n:
    lista_R[f"R{i}{j}"] = lista_R[f"R{i}{j-1}"] + ( (lista_R[f"R{i}{j-1}"] - lista_R[f"R{i-1}{j-1}"]) / (4**(j-1) - 1) )
    # Esta expresion de arriba es simplemente la ecuacion iterativa del metodo de Romberg
    print(i, j)
    i+=1
  j+=1
  print(j)  #se imprimen los indices para verificar que los indices actuen correctamente
```

# Definimos la funcion para un uso mas rapido y accesible

```python
def romberg(a,b,n,funcion):
  a = float(a)
  b = float(b)
  n = int(n)

  x = sp.Symbol('x') #asignamos la variable aleatoria X

  expresion = sp.sympify(funcion) #traducimos el texto para que programa pueda comprender que función se va a utilizar
  fx = sp.lambdify(x, expresion, 'math') #se define la función matemática

  lista_R = {
    "R11" : ((b-a)/2) * (fx(a) + fx(b))
  }

  def h(j):
    return (b-a) / 2**(float(j)-1)

  i=2

  while i<=n:
    lista_R[f"R{i}1"] = lista_R[f"R{i-1}1"] / 2
    sumatoria=0
    k=1
    while k <= 2**(i-2):
      sumatoria += fx(a + (2*k-1)* h(i))
      k+=1
    lista_R[f"R{i}1"] += 0.5 * h(i-1) * sumatoria
    i+=1

  i=2
  j=2 #Reiniciamos los contadores, ya que terminaron con el ultimo valor utlizado del cuadro de arriba

  while j<=n:
    i=j
    while i<=n:
      lista_R[f"R{i}{j}"] = lista_R[f"R{i}{j-1}"] + ( (lista_R[f"R{i}{j-1}"] - lista_R[f"R{i-1}{j-1}"]) / (4**(j-1) - 1) )
      i+=1
    j+=1

  return lista_R[f"R{i-1}{j-1}"]
```
