```python
import numpy as np
import matplotlib.pyplot as plt
```

Grafica de $f(x)=e^{x^2}$ y $P(x)=1+x^2+\frac{x^4}{3!}$

```python
def f(x):
  return np.exp(x**2)
```

```python
f(1)
```

```python
xx=np.linspace(0,1,100)
plt.plot(xx, f(xx))
```

```python
def P(x):
  return 1+(x**2)+(x**4/2)
```

```python
P(1)
```

```python
xx=np.linspace(0,1,100)
plt.plot(xx,f(xx),color='red')
plt.plot(xx,P(xx))
```
